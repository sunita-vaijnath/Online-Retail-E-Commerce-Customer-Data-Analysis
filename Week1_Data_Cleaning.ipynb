DATA QUALITY ASSESMENT

```python
negative_quantity_count = (df["Quantity"] < 0).sum()

print("Number of transactions with negative quantity:",
      negative_quantity_count)
```

```python
negative_quantity_rows = df[df["Quantity"] < 0]

negative_quantity_rows.head(10)
```

```python
negative_with_cancellation_invoice = (
    negative_quantity_rows["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

print(
    "Negative-quantity transactions with cancellation invoices:",
    negative_with_cancellation_invoice
)
```

```python
negative_without_cancellation = negative_quantity_rows[
    ~negative_quantity_rows["Invoice"]
    .astype(str)
    .str.startswith("C")
]

print(
    "Negative-quantity transactions without cancellation invoice:",
    len(negative_without_cancellation)
)
```

```python
negative_without_cancellation["Invoice"].astype(str).str[:1].value_counts()
```

```python
# 1. Count zero-quantity transactions
zero_quantity_count = (df["Quantity"] == 0).sum()

print("Number of transactions with zero quantity:",
      zero_quantity_count)


# 2. Extract zero-quantity transactions
zero_quantity_rows = df[df["Quantity"] == 0]

print("\nSample zero-quantity transactions:")
display(zero_quantity_rows.head(10))


# 3. Examine invoice-number patterns
invoice_patterns = (
    zero_quantity_rows["Invoice"]
    .astype(str)
    .str[:1]
    .value_counts()
)

print("\nInvoice-number patterns among zero-quantity transactions:")
print(invoice_patterns)
```

```python
# 1. Count negative-price transactions
negative_price_count = (df["Price"] < 0).sum()

print("Number of transactions with negative price:",
      negative_price_count)


# 2. Extract negative-price transactions
negative_price_rows = df[df["Price"] < 0]

print("\nSample negative-price transactions:")
display(negative_price_rows.head(10))


# 3. Examine invoice-number patterns
invoice_patterns = (
    negative_price_rows["Invoice"]
    .astype(str)
    .str[:1]
    .value_counts()
)

print("\nInvoice-number patterns among negative-price transactions:")
print(invoice_patterns)
```

```python

zero_price = df[df["Price"] == 0]

print("Zero-price transactions:", len(zero_price))
display(zero_price.head(10))
```

```python

missing_customer = df[df["Customer ID"].isna()]

print("Missing Customer ID:", len(missing_customer))
print("Percentage of dataset:",
      round(len(missing_customer) / len(df) * 100, 2), "%")

display(missing_customer.head(10))
```

```python
missing_description = df[df["Description"].isna()]

print("Missing Description:", len(missing_description))
print("Percentage of dataset:",
      round(len(missing_description) / len(df) * 100, 2), "%")

display(missing_description.head(10))
```

```python
duplicate_count = df.duplicated().sum()

print("Completely duplicated records:", duplicate_count)
print("Percentage of dataset:",
      round(duplicate_count / len(df) * 100, 2), "%")

display(df[df.duplicated(keep=False)].head(10))
```

```python
for column in ["Quantity", "Price"]:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\n{column}")
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

## DATA CLEANING

```python
# Create a copy of the raw dataset
df_clean = df.copy()

print("Original rows:", len(df_clean))
```

```python
# Remove exact duplicate records
df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
# Remove negative-price bad-debt adjustment records
df_clean = df_clean[df_clean["Price"] >= 0]

print("Rows after removing bad-debt records:", len(df_clean))
```

```python
# Remove zero-price transactions
df_clean = df_clean[df_clean["Price"] > 0]

print("Rows after removing zero-price transactions:", len(df_clean))
```

```python
# Fill missing product descriptions
df_clean["Description"] = df_clean["Description"].fillna("Unknown")

print("Missing descriptions:", df_clean["Description"].isna().sum())
```

```python
# Flag cancelled transactions
df_clean["IsCancellation"] = df_clean["Invoice"].astype(str).str.startswith("C")

print("Cancellation transactions:",
      df_clean["IsCancellation"].sum())
```

```python
# Final cleaning validation
print("Rows:", len(df_clean))
print("Duplicates:", df_clean.duplicated().sum())
print("Negative prices:", (df_clean["Price"] < 0).sum())
print("Zero prices:", (df_clean["Price"] == 0).sum())
print("Missing descriptions:", df_clean["Description"].isna().sum())
```

## PREPROCESSING

```python
# Calculate total transaction value
df_clean["TotalAmount"] = df_clean["Quantity"] * df_clean["Price"]

df_clean[["Quantity", "Price", "TotalAmount"]].head()
```

```python
# Create date-based features
df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["Day"] = df_clean["InvoiceDate"].dt.day

df_clean[["InvoiceDate", "Year", "Month", "Day"]].head()
```

```python
# Check final dataset structure
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print("\nMissing values:")
print(df_clean.isnull().sum())
```

```python
# Save cleaned dataset
df_clean.to_csv("/content/online_retail_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
